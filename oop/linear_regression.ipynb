```python
import matplotlib.pyplot as plt
```

```python
class LinearRegression:
    def __init__(self):
        self.m = 0
        self.b = 0
        self.learning_rate = 0.01

        self.mse = []
        self.slope_values = []
        self.intercept_values = []

        self.prediction_history = {}


    def fit(self, x:list, y:list, iterations):
        self.x = x
        self.y = y
        self.iterations = iterations

        self.slope_values.append(self.m)
        self.intercept_values.append(self.b)

        for iteration in range(self.iterations):
            print(f"Iteration {iteration}, Slope: {self.m}, Intercept: {self.b}")
            self.mse.append(self.mean_squared_error())
            self.prediction_history[iteration] = self.y_pred.copy()
            print(f"Loss {self.mse[-1]}")
            new_m = self.update_slope()
            new_b = self.update_intercept()
            self.m = new_m
            self.b = new_b
            self.slope_values.append(self.m)
            self.intercept_values.append(self.b)

        print(f"Slope values: {self.slope_values}")
        print(f"Intercept values: {self.intercept_values}")


    def predict(self, new_x:list):
        for x_value in new_x:
            x_value = float(x_value)
            return self.m * x_value + self.b


    def mean_squared_error(self):
        self.y_pred = []
        residuals_squared = []

        for x_index in range(len(self.x)):
            self.y_pred.append(self.m * self.x[x_index] + self.b)

        for y_index in range(len(self.y)):
            residuals_squared.append((self.y[y_index] - self.y_pred[y_index]) ** 2)

        return sum(residuals_squared) / len(self.y)


    def calculate_slope_gradient(self):
        sample_gradients = []
        for index in range(len(self.y)):
            sample_gradients.append((-2*self.x[index]*(self.y[index] - self.m * self.x[index] - self.b)))

        return sum(sample_gradients) / len(self.y)


    def calculate_intercept_gradient(self):
        sample_gradients = []
        for index in range(len(self.y)):
            sample_gradients.append((-2*(self.y[index] - self.m * self.x[index] - self.b)))

        return sum(sample_gradients) / len(self.y)


    def update_slope(self):
        return self.m - self.learning_rate * self.calculate_slope_gradient()


    def update_intercept(self):
        return self.b - self.learning_rate * self.calculate_intercept_gradient()
```

```python
x = [1,2,3,4,5]
y = [10,25,32,47,45]

model = LinearRegression()

model.fit(x, y, 20)
```

```python
model.predict([6])
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.scatter(x,y)
for iteration in range(model.iterations):
    plt.plot(x, model.prediction_history[iteration])
plt.title("Linear Regression")
plt.ylabel('Predicted Values')

plt.subplot(2,3,2)
for value in range(len(model.mse)):
    plt.scatter(model.slope_values[value], model.mse[value])
plt.title('Gradient Descent (Slope)')
plt.xlabel('Slope')
plt.ylabel('Mean Squared Error')

plt.subplot(2,3,3)
for value in range(len(model.mse)):
    plt.scatter(model.slope_values[value], model.mse[value])
plt.title('Gradient Descent (Intercept)')
plt.xlabel('Intercept')
plt.ylabel('Mean Squared Error')

plt.subplot(2,3,5)
for iteration in range(len(model.mse)):
    plt.scatter(iteration, model.mse[iteration])
plt.title('Loss History')
plt.xlabel('Iteration')
plt.ylabel('Loss')

plt.tight_layout()
plt.show()
```
