```python
import matplotlib.pyplot as plt
import numpy as np
from pure_python.functions import mean_squared_error, update_slope, update_intercept
```

```python
# Generate 10 random X values
x = np.random.randint(1, 20, size=10)

# Generate Y values with a linear relationship + random noise
y_true = 5 * x + 10 + np.random.normal(0, 5, size=10)
```

```python
# Initial predefined values for the slope and y-intercept.

m = 0 # slope
b = 0 # intercept
```

```python
# Empty lists for the slope, intercept, loss and prediction history.
# The values in slope and intercept will be used for the plots.

slope = []
intercept = []
loss_history = []
prediction_history = {}
```

```python
# Appending the initial slope and intercept values to the lists.

slope.append(m)
intercept.append(b)
```

```python
# The complete training process

for i in range (100):
    y_pred = []
    for value in range(len(x)):
        predicted_value = b + m * x[value]
        y_pred.append(predicted_value)

    prediction_history[i] = y_pred.copy()

    mse = mean_squared_error(y_true, y_pred)
    loss_history.append(mse)
    new_m = update_slope(x, y_true, m, b)
    new_b = update_intercept(x, y_true, m, b)

    m = new_m
    b = new_b

    slope.append(m)
    intercept.append(b)
    print(f" Iteration: {i}, Slope: {m}, Intercept: {b}, Loss: {loss_history[-1]} ")
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.scatter(x,y_true)
for iteration in prediction_history:
    plt.plot(x, prediction_history[iteration])
plt.title("Linear Regression")
plt.ylabel('Predicted Values')

plt.subplot(2,3,2)
for value in range(len(loss_history)):
    plt.scatter(slope[value], loss_history[value])

plt.title('Gradient Descent (Slope)')
plt.xlabel('Slope')
plt.ylabel('Mean Squared Error')

plt.subplot(2,3,3)
for value in range(len(loss_history)):
    plt.scatter(intercept[value], loss_history[value])
plt.title('Gradient Descent (Intercept)')
plt.xlabel('Intercept')
plt.ylabel('Mean Squared Error')

plt.subplot(2,3,5)
for iteration in range(len(loss_history)):
    plt.scatter(iteration, loss_history[iteration])
plt.title('Loss History')
plt.xlabel('Iteration')
plt.ylabel('Loss')

plt.tight_layout()
plt.show()
```
